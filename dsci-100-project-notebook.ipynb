#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
wmm_mmm <- wmm_tidy |>
    select(reg_wins, tourney_wins) |> #selecting numeric columns
    pivot_longer(cols = reg_wins:tourney_wins,
                 names_to = "type",
                 values_to = "wins") |> # pivoting the columns to rows so min, max, and mean can be columns
    group_by(type) |> #grouping to prevent min, max, and mean of all types instead of the groups
    summarize(wins_min = min(wins, na.rm = TRUE),
              wins_max = max(wins, na.rm = TRUE),
              wins_mean = mean(wins, na.rm = TRUE)) #making the min, max, and mean columns

wmm_mmm
```

```r
#This visualization will be based on how many schools have played based on the amount of time.
wmm_school <- wmm_tidy |> #making new data table 
    group_by(school) |> #grouping by schools
    summarize(count = n()) #counting the amount of times the schools have each played (in years)

wmm_school_plot <- wmm_school |> 
    ggplot(aes(x = count)) + #making the plot
    geom_histogram(bins = 20) + # using histogram for distribution 
    labs(x = "Amount of times played (in years)", y = "Number of Schools") + #labelling x as years played, y as the number of schools
    theme(text = element_text(size = 20)) #adjusting text size
wmm_school_plot
```

Visualization 1: The relationship between regional wins and school allows a school to be defined by its regional wins. Predicting which school is which will now have regional wins as its predictor. Since a school can play regionals many times, combining all the regional wins and taking the average of them will help us determine the regional win rate of a school.

```r
#since there are 284 schools in total, I will only be using the top 10 schools who have played the longest
wmm_10 <- wmm_tidy |>
    group_by(school) |> #grouping schools
    summarize(count = n(), #creating a new column for how many times the school has played
             reg_wins = mean(reg_wins, na.rm = TRUE)) |> #taking the mean of each school
    slice_max(count, n = 10) #taking the top ten longest playing schools
wmm_10

options(repr.plot.height = 6, repr.plot.width = 12)
wmm_reg_school_plot <- ggplot(wmm_10, aes(x = reg_wins, y = school)) + #setting the axises, putting the school on y for legibility
    geom_bar(stat = "identity") + #making a bar graph, setting stat as identity so the usual count can be replaced with reg_wins
    labs(x = "Regional Wins", y = "Top 10 Schools") + #labelling axises
    theme(text = element_text(size = 20)) #changing the text size for legiibility
wmm_reg_school_plot
```

Visualization 2: Using the relationship between tournament wins and school will also allow tournament wins to be a predictor of what school can be predicted to be. 

```r
wmm_10 <- wmm_tidy |>
    group_by(school) |> #doing the same thing as the previous visualization but for tourney_wins instead,
    summarize(count = n(),
             tourney_wins = mean(tourney_wins)) |> 
    slice_max(count, n = 10) #the schools have not changed as the schools are selected by count, not tourney_wins
wmm_10

options(repr.plot.height = 6, repr.plot.width = 12)
wmm_tourney_school_plot <- ggplot(wmm_10, aes(x = tourney_wins, y = school)) +
    geom_bar(stat = "identity") +
    labs(x = "Tournament Wins", y = "Top 10 Schools") +
    theme(text = element_text(size = 20))
wmm_tourney_school_plot
```
